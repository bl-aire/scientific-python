```python
# LIST FUNCTIONS
x = [1, 2, 3, 4, 5]
print(sum(x))

u = [0, 1, 2, 3]
v = ['a', 'b', 'c', 'd']
print(list(zip(u, v)))

a = [1, 0, 3, 5, 6, 9]
def less_than_5(x):
    """a function that returns True if x<5"""
    return x < 5
print(list(filter(less_than_5, a)))

def square(x):
    """a function that returns the square value"""
    return x**2
print(list(map(square, a)))

# Check if each item is less than 5
print(list(map(less_than_5, a)))

# Check if all elements are less than 5
print(all(list(map(less_than_5, a))))

# Check if any elements are less than 5
print(any(list(map(less_than_5, u))))
```

```python
# PYTHON IF STATEMENTS
"""
if <Boolean expression 1>:
    <block 1>
elif <Boolean expression 2>:
    <block 2>
else:
    <block 3>
<block 4>
"""

# LOOP CONSTRUCTS
# For loop
"""
for <iterator> in <iterable>:
    <block>
"""
x = 4
for x in "abc":
    print(x)
print(x)

Z = [('a', 1), ('b', 2), ('c', 3)]
for (x, y) in Z:
    print(x, y)

l = [0, 1, 2, 3, 4]
for i in range(len(l)):
    l[i] += 1
l

# A word of caution: The block can change the iterator and, in theory, also the iterable that is iterated over. Consider this simple example:
l = [0, 1, 2, 3, 4]
for i in range(len(l)):
    if i+1 < len(l):
        l[i+1] *= 2
    print(l[i])

# While loop
i = 0
while i < 5:
    print(i)
    i += 1
print('done!')

# Break and Consinue statemments
for i in range(10):
    if i > 5:
        break
    print(i)
print('done!')

for i in range(10):
    if i > 5:
        continue
    print(i)
print('done!')
```

```python
# ERROR HANDLING

# Traceback
def invert(p):
    return 1/p
invert(0)

# Errors, Exceptions, and Warnings
def invert_safe(m):
    try:
        return 1/m
    except ZeroDivisionError:
        return None
print(invert_safe(0))
print(invert_safe(2))

x = 6
if x >= 5:
    raise ValueError('x must be less than 5 (f g provided)'.format(x))
```

```python
# PYTHON CLASSES
class Frac:
    """Fractional class, consisting of a pair of integers."""
    def init (self, num, den):
        """Constructor; catch case denominator == 0."""
        self.num = num
        if den == 0:
            raise ValueError("Denominator cannot be zero.")
        self.den = den
        
    def repr (self):
        """Generate a string representation of a Frac. """
        return "f:dg/f:dg".format(self.num, self.den)
        
    def mul (self, another):
        """Multiply two Fracs to produce a Frac. """
        return Frac(self.num*another.num, self.den*another.den)

    def add (self, another):
        """Add two Fracs to produce a Frac. """
        return Frac(self.num*another.den+self.den*another.num, self.den*another.den)

    def reduce(self):
        """Reduce numerator and denominator based on the greatest
        common divisor, derived with the Euclidean algorithm."""
        a, b = self.num, self.den
        while b != 0:
            (a, b) = (b, a%b)
        self.num = int(self.num/a)
        self.den = int(self.den/a)
        return self

    def to real(self):
        """ Return floating point value of Frac. """
        return float(self.num)/float(self.den)
```
