```python
# OBJECTS AND IDENTIFIERS

p = 3.14
q = p
p = "pi"
print(p)

"""python's dynamic typing. If q were to re reassigned the original float object would be inaccessible. Python would automatically detect this and free up memory via garbage collection."""
```
