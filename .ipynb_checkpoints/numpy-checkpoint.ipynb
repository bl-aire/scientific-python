```python
# NumPy: NUMERICAL MATHEMATIC

import numpy as np

# Numpy array efficiency

import time

a = list(range(10000000))
b = []
starttime = time.time()
for i in range(len(a)):
    b.append(a[i]+1)
print(time.time()-starttime, 'seconds')

c = np.array(a)
starttime = time.time()
d = c + 1
print(time.time()-starttime, 'seconds')

# One-dimensional arrays
l = [1, 2, 3.0]
x = np.array(l)
print(x)

# Specify datatype
np.array(l, dtype=complex)

# Ndarray Attributes
print(x.dtype)
print(x.ndim)
print(x.size)
print(x.shape)

# Ndarray Methods
print(x.astype(int))

x = np.linspace(0, 10, 11)
print(x)
print(x.mean()) # mean of array
print(x.max()) # max of array
print(x.argmax()) # index of max element
print(x.clip(2, 6)) # clip array values to range [2,6]
```
